```python
from sklearn.datasets import fetch_california_housing
from sklearn.metrics import root_mean_squared_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
```

```python
#MLPReegeressor
housing = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(
housing.data, housing.target, random_state=42)
mlp_reg = MLPRegressor(hidden_layer_sizes=[50, 50, 50], early_stopping=True,
verbose=True, random_state=45,activation = 'tanh',learning_rate_init= 0.002)
```

```python
pipeline = make_pipeline(StandardScaler(),mlp_reg)
pipeline.fit(X_train,y_train)
```

```python
 mlp_reg.best_validation_score_
```

```python
y_pred = pipeline.predict(X_test)
rmse = root_mean_squared_error(y_test, y_pred)
rmse
```

```python
#MLPClassifier
from sklearn.datasets import fetch_openml
fashion_mnist = fetch_openml(name = "Fashion-MNIST",as_frame = False)
```

```python
targets = fashion_mnist.target.astype(int)
X_train1, y_train1 = fashion_mnist.data[:60_000], targets[:60_000]
X_test1, y_test1 = fashion_mnist.data[60_000:], targets[60_000:]
```

```python

X_sample = X_train1[1].reshape(28, 28)  
plt.imshow(X_sample, cmap="binary")
plt.show()
```

```python
class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
"Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import MinMaxScaler
```

```python
import time
mlp_clf = MLPClassifier(hidden_layer_sizes = [300,100],verbose = True,early_stopping = True,random_state = 46)
pipeline = make_pipeline(MinMaxScaler(), mlp_clf)
start = time.perf_counter()
pipeline.fit(X_train1, y_train1)
end = time.perf_counter()
accuracy = pipeline.score(X_test1, y_test1)
```

```python
print(f"Training time: {end - start:.6f} seconds")
accuracy
```
