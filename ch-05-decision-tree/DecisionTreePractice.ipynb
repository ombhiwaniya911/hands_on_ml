```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier as DT
iris = load_iris(as_frame = True)
x = iris.data[["petal length (cm)","petal width (cm)"]].values
y = iris.target
tree_clf = DT(max_depth = 2,random_state = 32)
tree_clf.fit(x,y)
```

```python
from sklearn.tree import export_graphviz
export_graphviz(
tree_clf,
out_file="iris_tree.dot",
feature_names=["petal length (cm)", "petal width (cm)"],
class_names=iris.target_names,
rounded=True,
filled=True
)
```

```python
from graphviz import Source
Source.from_file("iris_tree.dot")
```

```python
from sklearn.datasets import make_moons
X_moons, y_moons = make_moons(n_samples=150, noise=0.2, random_state=42)
tree_clf1 = DT(random_state=42)
tree_clf2 = DT(min_samples_leaf=5, random_state=42)
tree_clf1.fit(X_moons, y_moons)
tree_clf2.fit(X_moons, y_moons)
```

```python
X_moons_test, y_moons_test = make_moons(n_samples=1000, noise=0.2,                                        
random_state=43)

print(tree_clf1.score(X_moons_test, y_moons_test))

print(tree_clf2.score(X_moons_test, y_moons_test))

```

```python
import numpy as np
from sklearn.tree import DecisionTreeRegressor
rng = np.random.default_rng(seed=42)
X_quad = rng.random((200, 1)) - 0.5  # a single random input feature
y_quad = X_quad ** 2 + 0.025 * rng.standard_normal((200, 1))
tree_reg = DecisionTreeRegressor(max_depth=2, random_state=42)
tree_reg.fit(X_quad, y_quad)
```

```python
export_graphviz(
tree_reg,
out_file="regressor.dot",
feature_names=["X_quad"],
rounded=True,
filled=True
)
```

```python
Source.from_file("regressor.dot")
```
